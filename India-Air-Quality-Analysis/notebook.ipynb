```python
def get_aqi_category(aqi):
    if aqi <= 50:
        return "Good"
    elif aqi <= 100:
        return "Satisfactory"
    elif aqi <= 200:
        return "Moderate"
    elif aqi <= 300:
        return "Poor"
    elif aqi <= 400:
        return "Very Poor"
    else:
        return "Severe"
```

```python
df["calculated_aqi_category"] = (
    df["aqi"].apply(get_aqi_category)
)
```

```python
df["quarter"] = df["date"].dt.quarter
```

```python
def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Summer"
    elif month in [6, 7, 8, 9]:
        return "Monsoon"
    else:
        return "Post-Monsoon"

df["season"] = df["month"].apply(get_season)
```

```python
like_for_like_annual = (
    long_term_df.groupby("year")
    .agg(
        avg_aqi=("aqi", "mean"),
        median_aqi=("aqi", "median"),
        cities=("city", "nunique"),
        observations=("aqi", "count")
    )
    .reset_index()
)

like_for_like_annual
```

```python
comparison = annual_aqi[
    ["year", "avg_aqi"]
].rename(
    columns={"avg_aqi": "all_cities_aqi"}
).merge(
    like_for_like_annual[
        ["year", "avg_aqi"]
    ].rename(
        columns={"avg_aqi": "long_term_cities_aqi"}
    ),
    on="year",
    how="inner"
)

comparison
```

```python
comparison_long = comparison.melt(
    id_vars="year",
    value_vars=[
        "all_cities_aqi",
        "long_term_cities_aqi"
    ],
    var_name="series",
    value_name="avg_aqi"
)

comparison_long["series"] = comparison_long["series"].replace({
    "all_cities_aqi": "All Monitored Cities",
    "long_term_cities_aqi": "Long-Term City Cohort"
})
```

```python
fig = px.line(
    comparison_long,
    x="year",
    y="avg_aqi",
    color="series",
    markers=True,
    title="Overall Monitoring Network vs Like-for-Like AQI Trend"
)

fig.update_layout(
    xaxis_title="Year",
    yaxis_title="Average AQI",
    legend_title=""
)

fig.show()
```

```python
episode_df = (
    df_clean[
        ["date", "city", "aqi", "aqi_category", "prominent_pollutant"]
    ]
    .sort_values(["city", "date"])
    .copy()
)
```

```python
episode_df["high_pollution"] = (
    episode_df["aqi"] > 300
).astype(int)
```

```python
episode_df["prev_date"] = (
    episode_df.groupby("city")["date"]
    .shift()
)

episode_df["prev_high"] = (
    episode_df.groupby("city")["high_pollution"]
    .shift(fill_value=0)
)
```

```python
episode_df["new_episode"] = (
    (episode_df["high_pollution"] == 1)
    &
    (
        (episode_df["prev_high"] == 0)
        |
        ((episode_df["date"] - episode_df["prev_date"]).dt.days > 1)
    )
).astype(int)
```

```python
episode_df["episode_id"] = (
    episode_df.groupby("city")["new_episode"]
    .cumsum()
)
```

```python
high_pollution_df = episode_df[
    episode_df["high_pollution"] == 1
].copy()
```

```python
pollution_episodes = (
    high_pollution_df.groupby(
        ["city", "episode_id"]
    )
    .agg(
        start_date=("date", "min"),
        end_date=("date", "max"),
        duration_days=("date", "count"),
        avg_aqi=("aqi", "mean"),
        peak_aqi=("aqi", "max")
    )
    .reset_index()
)
```

```python
pollution_episodes = pollution_episodes[
    pollution_episodes["duration_days"] >= 2
].copy()
```

```python
longest_episodes = (
    pollution_episodes
    .sort_values(
        ["duration_days", "peak_aqi"],
        ascending=[False, False]
    )
    .head(15)
)

longest_episodes
```
